```python
def fib(n):
    if n <= 1:
        return n
    else:
        return fib(n-1) + fib(n-2)
```

```python
fib(40)
```

```python
def fib_memo(n, memo={}):
  if n in memo:
    return memo[n]
  if n <= 1:
    return n
  memo[n] = fib_memo(n - 1, memo) + fib_memo(n - 2, memo)
  return memo[n]
```

```python
fib_memo(40)
```

```python
def fib_tab(n):
  if n <= 1:
    return n
  prev2, prev1 = 0, 1
  for _ in range(2, n + 1):
    prev2, prev1 = prev1, prev2 + prev1
  return prev1
```

```python
fib_memo(40)
```
